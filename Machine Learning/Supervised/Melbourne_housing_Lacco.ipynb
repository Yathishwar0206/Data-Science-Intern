## Importing Liabraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

## Loading the Dataset

```python
data = pd.read_csv(r'C:\Users\ADMIN\Desktop\Jupyter folder/Downloads/Melbourne_housing_FULL.csv')
data
```

```python
data.shape
```

```python
data.head()
```

```python
data.tail()
```

```python
data.nunique()
```

```python
coln = ['Suburb','Rooms','Type','Method','SellerG','Car','Bathroom','Bedroom2','Landsize','Distance','CouncilArea','BuildingArea','Regionname',
       'Propertycount','Price']
data = data[coln]
data
```

```python
data.shape
```

## Checking of Null values

```python
data.isnull().sum()
```

## Handling the Missing values

```python
ctfz = ['Propertycount','Distance','Car','Bathroom','Bedroom2']
data[ctfz] = data[ctfz].fillna(0)

data['Landsize'] = data['Landsize'].fillna(data.Landsize.mean())
data['BuildingArea'] = data['BuildingArea'].fillna(data.BuildingArea.mean())
```

```python
data.dropna(inplace=True)
data
```

```python
data.shape
```

## Let's One hot encode the categorical Features

```python
data = pd.get_dummies(data, drop_first=True)
data.head()
```

## Split into two

```python
x = data.drop('Price',axis=1)
y = data['Price']
x
```

```python
y
```

## Train and Test

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.3, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
rg = LinearRegression().fit(X_train, y_train)
```

```python
rg.score(X_test, y_test)
```

```python
rg.score(X_train, y_train)
```

Here the Training Score is 68% and the Testing Score is 13 %.

<h3 style='color:blue', align='center'> L1 Regularization </h3>

## Using Lasso(L1) regularization model

```python
from sklearn import linear_model
lasso_reg = linear_model.Lasso(alpha=50 ,max_iter=100, tol=0.1)
lasso_reg.fit(X_train, y_train)
```

```python
lasso_reg.score(X_test, y_test)
```

```python
lasso_reg.score(X_train, y_train)
```

## Using Ridge (L2) Reguralization model

```python
from sklearn import linear_model
ridge_reg = linear_model.Ridge(alpha=50, max_iter=100, tol=0.1)
ridge_reg.fit(X_train, y_train)
```

```python
ridge_reg.score(X_test, y_test)
```

```python
ridge_reg.score(X_train, y_train)
```
